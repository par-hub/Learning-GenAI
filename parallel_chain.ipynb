```python
import os
from langchain_core.prompts import PromptTemplate
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.runnables import RunnableParallel
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv()
```

```python
prompt1 = PromptTemplate(template="Generate a summary of the folowing topic: {topic}", input_variables=["topic"])
```

```python
prompt2 = PromptTemplate(template="Generate a 5 Question quiz on this {topic}", input_variables=["topic"])
```

```python
MODEL1 = "openai/gpt-oss-120b"
MODEL2 = "openai/gpt-oss-20b"
model = ChatGroq(
    model=MODEL1,
    temperature=0,
)
model1 = ChatGroq(
    model=MODEL2,
    temperature=0,
)
```

```python
parser = StrOutputParser()
```

```python
prallel_chain=RunnableParallel({'summary': prompt1 | model | parser, 'quiz': prompt2 | model1 | parser})
```

```python
prompt3 = PromptTemplate(template = 'Merge the following summary and quiz together: {summary} {quiz}', input_variables=['summary', 'quiz'])
```

```python
chain3 = prompt3 | model | parser
```

```python
final_chain = prallel_chain | chain3
```

```python
response = final_chain.invoke({"topic": "Generative AI"})
```
