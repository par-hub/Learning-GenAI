```python
from pydantic import BaseModel,Field  # Base model is responsible for creating the response as the way we want 
from typing import Optional, List
```

```python


class Review(BaseModel):
    Product: str = "Not provided"
    pros: list[str]
    cons: list[str]
    Reviwer: Optional[str] = "Anonymous"
    Rating: Optional[float] = Field(default=3, ge=0, le=5, description="Rating should be between 0 and 5")
```
