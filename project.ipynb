```python
inputi =tokenizer("write a python code to print the sum of two numbers",return_tensors="pt")
```

```python
model = AutoModelForCausalLM.from_pretrained("google/gemma-3-1b-it",dtype='bfloat16')
```

```python
out = model(input_ids=inputi['input_ids'])
```

```python
gen_out = model.generate(input_ids=inputi['input_ids'],max_new_tokens=300)
```

```python
gen_out
```

```python
tokenizer.batch_decode(gen_out, skip_special_tokens=True)
```
