```python
%pip install -U langchain-google-genai
from langchain_ollama import ChatOllama
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
```

```python
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
google_genai = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
ollama = ChatOllama(model="gemma3:latest")
```

```python
promt1 = PromptTemplate(template="Generate a Detailed report on the following {Topic}", input_variables=["Topic"])
```

```python
promt2 = PromptTemplate(template="Generate a 5 point summary on the {Report}", input_variables=["Report"])
```

```python
final1 = promt1.invoke({"Topic": "Google vs ollama models"})
```

```python
parser= StrOutputParser()
```

```python
chain = promt1 | google_genai |parser |promt2 | ollama | parser
```

```python
response = chain.invoke({"Topic": "Google vs ollama models"})
```
