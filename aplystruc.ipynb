```python
from pydantic import BaseModel, Field
from typing import Optional, Literal
from langchain_ollama import ChatOllama
```

```python
class Data(BaseModel):
    product_name: str = Field(description="Provied the name of the product")
    pros: list[str] = Field(description="Mention the positive aspects of the product")
    cons: list[str] = Field(description="Mention the negative aspects of the product")
    sentiment: Literal["positive", "negative", "neutral"] = Field(description="The sentiment of the review")
    rating: float = Field(ge=1,le=5,default=3,description="The rating of the product out of 5")
```

```python
model = ChatOllama(model="gemma3:latest")
```

```python
structure = model.with_structured_output(Data)
```

```python
print = structure.invoke(Review)
```
